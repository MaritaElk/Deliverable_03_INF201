```python
from mage import Mage

mage = Mage("Gandalf")

print(mage)
print(mage.mana)
print(mage.mana_max)
```

```python
from character import Character
from mage import Mage

gandalf = Mage("Gandalf")
enemy = Character("Enemy", hp_max=100)

gandalf.attack(enemy)

print(enemy.hp)
print(gandalf.mana)
```

```python
gandalf = Mage("Gandalf")
enemy = Character("Enemy", hp_max=100)

gandalf.mana = 5

gandalf.attack(enemy)

print(enemy.hp)
print(gandalf.mana)
```

```python
gandalf = Mage("Gandalf")

gandalf.mana = 28
gandalf.attack(Character("Enemy"))

print(gandalf.mana)
```

```python
gandalf = Mage("Gandalf")

gandalf.mana = 28
gandalf.mana = min(gandalf.mana + 5, gandalf.mana_max)

print(gandalf.mana)
```

```python
gandalf = Mage("Gandalf", mana_max=12)
enemy = Character("Enemy", hp_max=100)

gandalf.mana = 9

gandalf.attack(enemy)

print(enemy.hp)
print(gandalf.mana)
```
